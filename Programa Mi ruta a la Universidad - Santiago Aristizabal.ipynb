```python
#Variables
codigo1 = "A00432291"
codigo2 = "A00130109"
#Variables para el calculo del costo del desplazamiento
tarifa_base = 5000
costo_por_km = 1200
cargo_hora_pico = 3000
margen_seguridad =(10) #En minutos.

#Impacto ambiental g CO2/hm
caminata = 0
bicicleta = 0
transporte_publico = 80
motocicleta = 100
automovil = 180

#Entrada previa al menu
print("-----------------------------Bienvenido a TU RUTA A LA U-----------------------------")
print("Para acceder al sistema, ingresa tu usuario: ")
usuario = input("Ingresa tu usuario: ")

if usuario == codigo1 or usuario == codigo2:
  print("¡Hola!", usuario)
  print("Bienvenido al sistema. Selecciona una opción del menú para continuar:")
  print("\n\n\n --------------MENU PRNCIPAL---------------")
  print("1. Calcular el costo estimado del desplazamiento\n2. Analizar las características del desplazamiento\n3. Recomendar un medio de transporte\n4. Determinar si llegará a tiempo a la universidad\n5. Estimar el impacto ambiental del desplazamiento\n6. Salir")

  opcion = int(input("Selecciona una opción: "))
  
  if opcion == 1:
    print("has seleccionado la opcion 1: calcular el costo estimado del desplazamiento")
    distancia = float(input("Ingresa la distancia entre el lugar de origen a la universidad en kilometros: "))
    hora_pico = int(input("¿Tienes planeado viajar en hora pico? 0. No / 1. Si :"))
    if hora_pico != 1 and hora_pico != 0:
      mensaje1 = ("Ingrese un valor que sea valido")
      print(mensaje1)
    else:        
      costo_total = (tarifa_base + (costo_por_km * distancia) + (cargo_hora_pico * hora_pico))
      print("EL costo total estimado del desplazamiento es de: ", costo_total , "pesos colombianos")
  
  elif opcion == 2:
    print("\nHas seleccionado la opcion 2: Analizar las características del desplazamiento")
    distancia = float(input("Ingresa la distancia a la que te encuentas de la universidad en kilometros:"))
    print("Distancia = " + str(distancia) + " km")
    if distancia < 0:
      print("Distancia NO valida")
    elif distancia <= 2:
      print("Distancia: Trayecto cercano")
    elif distancia <= 8:
      print("Distancia: Trayecto intermedio")
    else:
      print("Distancia: Trayecto distante")
    tiempo = float(input("Ingrese el tiempo estimado de desplazamiento en minutos:"))
    print("Tiempo Estimado = " + str(tiempo) + " minutos")
    if tiempo < 30: 
      print("Tiempo: Desplazamiento rápido")
    elif tiempo <= 60:
      print("Tiempo: Desplazamiento moderado")
    else:
      print("Tiempo: Desplazamiento prolongado")
    if distancia <= 2 and tiempo > 30:
      print("El tiempo de desplazamiento es alto en relación con la distancia. Conviene revisar la ruta utilizada.")

  elif opcion == 3:
    print("\nHas seleccionado la opción 3: Recomendar un medio de transporte")
    distancia3 = float(input("\nIngrese la distancia a la que se encuentra de la universidad en kilometros: "))
    if distancia3 < 0:
      print("Ingrese una distancia que sea valida.")
    elif distancia3 <= 2:
      lluvia = input("¿Esta lloviendo? (si/no): ").lower()
      if lluvia == "no":
        print("Se recomienda que camine a la universidad")
      else:
        print("Se recomienda ir en transporte publico a la universidad")
    elif distancia3 <= 8:
      lluvia = input("¿Esta lloviendo? (si/no):").lower()
      tiene_bicicleta = input("¿Tienes una bicicleta? (si/no): ").lower()
      if lluvia == "no" and tiene_bicicleta == "si": 
        print("Se recomienda ir en bicicleta a la universidad")
      else:
        print("Se recommiennda ir en transporte publico a la universidad")
    else: 
      print("Se recomienda ir en transporte publico o vehiculo compartido a la universidad")

  elif opcion == 4: 
    print("\nHas seleccionado la opción 4: Determinar si llegará a tiempo a la universidad")
    minutos_disponibles =float(input("¿cuantos minutos tienes disponibles antes de la clase?"))
    print("Posees " + str(minutos_disponibles) + " minutos disponibles")
    minutos_necesario = float(input("¿Cuantos minutos demoras de tu ubicación a la universidad aproximadamente?"))
    tiempo_necesario = minutos_necesario + margen_seguridad
    print("El tiempo nececesario sera = ",tiempo_necesario, "minutos")
    if (minutos_disponibles >= tiempo_necesario):
      print("Puedes realizar el desplazamiento y llegar a tiempo a la universidad")
    else: 
      print("Existe riesgo de llegar tarde. Se recomienda salir antes o considerar otra alternativa de transporte.")

  elif opcion == 5:
    print("\nHas seleccionado la opción 5: Estimar el impacto ambiental del desplazamiento:")
    distancia5 = float(input("Ingrese la cantidad de kilometros recorridos:"))
    print("\nLa cantidad de kilometros recorridos fue: ", distancia5)
    print("\nSeleccione una de las siguientes opciones: ")
    print("\n1.Caminata\n2.Bicicleta\n3.Transporte publico\n4.Motocicleta\n5.Automóvil")
    medio_transporte = int(input("Ingrese el medio de transporte que utilizo: "))
    
    if medio_transporte == 1:
      resultado = (distancia5 * caminata)
      print("\nLa emision de gramos de CO2 que usted ha hecho es = ", resultado)
    elif medio_transporte == 2:
      resultado = (distancia5 * bicicleta)
      print("\nLa emisión de gramos de CO2 que usted ha hecho es = ", resultado)
    elif medio_transporte == 3:
      resultado = (distancia5 * transporte_publico)
      print("\nLa emisión de gramos de CO2 que usted ha hecho es = ", resultado)
    elif medio_transporte == 4:
      resultado = (distancia5 * motocicleta)
      print("\nLa emisión de gramos de CO2 que usted ha hecho es = ", resultado)
    elif medio_transporte == 5:
      resultado = (distancia5 * automovil)
      print("\nLa emisión de gramos de CO2 que usted ha hecho es = ", resultado)
    else:
      resultado = None
      print("\nSeleccione una opción que si sea valida")
    if resultado is not None:
      if resultado == 0:
        print("Clasificación: Sin emisiones directas")
      elif resultado <= 500:
        print("Clasificación: Impacto bajo")
      elif resultado <= 1500:
        print("Clasificación: Impacto moderado")
      else:
        print("Clasificación: Impacto alto")

  elif opcion == 6:
    print("\nSe da por finalizada la ejecución del pograma.\nGracias por hacer uso de el.")
else: #Sale del sistema
   print("Acceso denegado \nCodigo no reconocido.")  
```
